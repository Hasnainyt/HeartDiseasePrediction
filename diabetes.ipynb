```python
# Importing dependencies
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn import svm
from sklearn.metrics import accuracy_score

# ----------------------------------------------------
# Data Collection & Analysis
# ----------------------------------------------------

# Load dataset
diabetes_dataset = pd.read_csv('diabetes.csv')

# Optional EDA
print(diabetes_dataset.head())
print(diabetes_dataset.shape)
print(diabetes_dataset.describe())
print(diabetes_dataset['Outcome'].value_counts())
print(diabetes_dataset.groupby('Outcome').mean())

# ----------------------------------------------------
# Separating features and target
# ----------------------------------------------------

X = diabetes_dataset.drop(columns='Outcome', axis=1)
y = diabetes_dataset['Outcome']

# ----------------------------------------------------
# Train-Test Split (IMPORTANT: before scaling)
# ----------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=1
)

# ----------------------------------------------------
# Data Standardization (CORRECT WAY)
# ----------------------------------------------------

scaler = StandardScaler()

# Fit ONLY on training data
X_train = scaler.fit_transform(X_train)

# Transform test data using same scaler
X_test = scaler.transform(X_test)

# ----------------------------------------------------
# Training the Model
# ----------------------------------------------------

classifier = svm.SVC(kernel='linear')
classifier.fit(X_train, y_train)

# ----------------------------------------------------
# Model Evaluation
# ----------------------------------------------------

# Training accuracy
X_train_prediction = classifier.predict(X_train)
train_accuracy = accuracy_score(X_train_prediction, y_train)
print("Training Accuracy:", train_accuracy)

# Test accuracy
X_test_prediction = classifier.predict(X_test)
test_accuracy = accuracy_score(X_test_prediction, y_test)
print("Test Accuracy:", test_accuracy)

# ----------------------------------------------------
# Predictive System
# ----------------------------------------------------

input_data = (7, 196, 90, 0, 0, 39.8, 0.451, 41)

# Convert input data to numpy array
input_array = np.asarray(input_data)

# Reshape data (1 row, 8 columns)
input_reshaped = input_array.reshape(1, -1)

# Standardize input using SAME scaler
std_input = scaler.transform(input_reshaped)

# Prediction
prediction = classifier.predict(std_input)
print("Prediction:", prediction)

if prediction[0] == 0:
    print("Not Diabetic Woman")
else:
    print("Diabetic Woman")
```

```python
import joblib

joblib.dump(classifier, "diabetes_model.pkl")
joblib.dump(scaler, "scaler.pkl")
```
