```python
# Importing dependencies
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn import svm
from sklearn.metrics import accuracy_score

# ----------------------------------------------------
# Data Collection & processing
# ----------------------------------------------------

# Load dataset
heart_data = pd.read_csv('heart.csv')


print(heart_data.head())
print(heart_data.shape)
#checking missing values
heart_data.isnull().sum()

print(heart_data.describe())
print(heart_data['target'].value_counts())
#1 = not healthy
#0=healthy
print(heart_data.groupby('target').mean())

# ----------------------------------------------------
# Separating features and target
# ----------------------------------------------------

X = heart_data.drop(columns='target', axis=1)
y = heart_data['target']

# ----------------------------------------------------
# Train-Test Split (IMPORTANT: before scaling)
# ----------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=2
)

# ----------------------------------------------------
# Data Standardization
# ----------------------------------------------------

scaler = StandardScaler()

# Fit ONLY on training data
X_train = scaler.fit_transform(X_train)

# Transform test data using same scaler
X_test = scaler.transform(X_test)

# ----------------------------------------------------
# Training the Model
# ----------------------------------------------------
# svm
classifier = svm.SVC(kernel='linear')
classifier.fit(X_train, y_train)

#logistic regression
model=LogisticRegression()
model.fit(X_train,y_train)

# ----------------------------------------------------
# Model Evaluations
# ----------------------------------------------------

# Training accuracy(svm)
X_train_prediction = classifier.predict(X_train)
train_accuracy = accuracy_score(X_train_prediction, y_train)
print("Training Accuracy of support vector machine:", train_accuracy)

#training accuracy(logistic)
X_train_prediction1 = model.predict(X_train)
train_accuracy1 = accuracy_score(X_train_prediction1, y_train)
print("Training Accuracy of logistic regression:", train_accuracy1)

# Test accuracy of svm
X_test_prediction = classifier.predict(X_test)
test_accuracy = accuracy_score(X_test_prediction, y_test)
print("Test Accuracy of svm:", test_accuracy)

# Test accuracy  of logistic
X_test_prediction1 = model.predict(X_test)
test_accuracy1 = accuracy_score(X_test_prediction1, y_test)
print("Test Accuracy of logistic regression:", test_accuracy1)

# ----------------------------------------------------
# Predictive System
# ----------------------------------------------------

input_data = (67,0,0,106,223,0,1,142,0,0.3,2,2,2)

# Column names same as training data
columns = ['age','sex','cp','trestbps','chol','fbs','restecg',
           'thalach','exang','oldpeak','slope','ca','thal']

# Create a DataFrame
input_df = pd.DataFrame([input_data], columns=columns)
# Convert input data to numpy array
input_array = np.asarray(input_data)

# Reshape data (1 row, 14 columns)
input_reshaped = input_array.reshape(1, -1)

# Standardize input using SAME scaler
std_input = scaler.transform(input_reshaped)

# Predictions
prediction = classifier.predict(std_input)
print("Prediction of svm:", prediction)

prediction1 = model.predict(std_input)
print("Prediction of logistic:", prediction1)

results = pd.DataFrame({
    "Model": ["SVM", "Logistic Regression"],
    "Prediction": [prediction[0], prediction1[0]],
    "Heart Condition": ["Healthy" if prediction[0]==0 else "Non Healthy",
                        "Healthy" if prediction1[0]==0 else "Non Healthy"],
    "Accuracy":[test_accuracy,test_accuracy1]
})
results

```
